```python
import tensorflow as tf
from tensorflow import keras
```

```python
(X_train,y_train),(X_test,y_test) =keras.datasets.mnist.load_data()
#previously you were using (X_train,X_test) but here this is the
 #way to do
 #X-train is having 60000 images
```

```python
X_train=X_train/255.0
X_test=X_test/255.0
#Normalizing the data to get in range between 0 10 1
#as 0 is white and 255 is black, so we divide i by 255
```

```python
model=keras.Sequential([
    keras.layers.Flatten(input_shape=(28,28)),
    keras.layers.Dense(128,activation='relu'),
    keras.layers.Dense(10,activation='softmax')

])
model.summary()
#here, 10 refers to output
#flatten: to convert multi dimension to single dimensio, and it retains the original copy and gives the copied
#version of it
#softmax: when there is multiple classes in the output
#param is the calculation, since no calculation occurs here flatten_4 so param is 0
```

```python
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy']
              )
#here we are using sparse_categorical_crossentropy is used because
#the output is 10 layers which doesn't have any binary values(2 class problem)
#so, it's considered as multi class problem
```

```python
model.fit(X_train,y_train,epochs=10, validation_data=(X_test,y_test))
```

```python
model.evaluate(X_test,y_test)
```

```python
loss, accuracy = model.evaluate(X_test, y_test)
print("Test Loss: ", loss)
print("Test Accuracy: ", accuracy)
```

```python
#take one test image
sample_img= X_test[0]
sample_label= y_test[0]

```

```python
sample_img = np.expand_dims(sample_img, axis=0) #(1,28,28)
sample_img.shape
```

```python
#predict
pred = model.predict(sample_img)
pred_digit = np.argmax(pred)
print("Predicted digit: ", pred_digit)
print("Actual digit: ", sample_label)
```
